```python
path_3 = "./data/mexico-real-estate-3.csv"
df3 = (
    pd.read_csv(path_3)
    .dropna()
)
df3
```

```python
df3.dtypes
```

```python
df3 = df3.assign(
    lat = lambda x : x["lat-lon"].str.split(",", expand=True)[0].astype(float),
    lon = lambda x : x["lat-lon"].str.split(",", expand=True)[1].astype(float),
    state = lambda x : x["place_with_parent_names"].str.split("|", expand=True)[2]
)
df3
```

```python
df3.dtypes
```

```python
df3 = df3.drop(columns=["place_with_parent_names","lat-lon"])
df3
```

```python
df3.shape
```

```python
# Check if all column sets are identical
columns_match = set(df1.columns) == set(df2.columns) == set(df3.columns)

if columns_match:
    print("✅ All DataFrames have the same columns.")
else:
    print("❌ Column mismatch detected!")
    # Optional: See the difference
    print("df1 vs df2 diff:", set(df1.columns) ^ set(df2.columns))
    print("df2 vs df3 diff:", set(df2.columns) ^ set(df3.columns))
```

```python
df = pd.concat([df1,df2,df3], ignore_index=True)
print(f"Combined shape: {df.shape}")
df.head()
```

```python
if df.shape[0] == ((df1.shape[0]) + (df2.shape[0]) + (df3.shape[0])):
    print("✅ All DataFrames have the same columns.")
else:
    print("❌ All DataFrames dont have the same columns.")
```
